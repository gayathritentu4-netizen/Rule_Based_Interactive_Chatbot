```python
import re
import json

print("Libraries imported successfully!")
```

```python
with open("sample_faq.json", "r") as file:
    faq_data = json.load(file)

print("FAQ dataset loaded successfully!")
```

```python
def normalize_input(user_input):

    user_input = user_input.lower().strip()

    user_input = re.sub(r'[^\w\s]', '', user_input)

    return user_input
```

```python
message = "  HELLO!!! "

result = normalize_input(message)

print("Original message:", message)
print("Normalized message:", result)
```

```python
print(normalize_input("GOOD MORNING!!"))
print(normalize_input("  How are you?  "))
print(normalize_input("BYE!!!"))
```

```python
def is_greeting(user_input):

    message = normalize_input(user_input)

    for greeting in faq_data["greetings"]:

        pattern = r'\b' + re.escape(greeting) + r'\b'

        if re.search(pattern, message):
            return True

    return False
```

```python
print(is_greeting("Hello"))
print(is_greeting("HI"))
print(is_greeting("Good morning"))
print(is_greeting("How are you?"))
```

```python
faq_patterns = {

    0: r'\b(hours?|operation|timings?|open|close)\b',

    1: r'\b(submit|submission|internship|project|github|repository)\b',

    2: r'\b(certificate|certificates|receive|issued)\b'

}

print("FAQ patterns created successfully!")
```

```python
message = "What are your working hours?"

for index, pattern in faq_patterns.items():

    if re.search(pattern, normalize_input(message)):
        print("Matched FAQ:", index)
```

```python
message = "How do I submit my internship project?"

for index, pattern in faq_patterns.items():

    if re.search(pattern, normalize_input(message)):
        print("Matched FAQ:", index)
```

```python
def recognize_intents(user_input):

    message = normalize_input(user_input)

    matched_faqs = []

    for index, pattern in faq_patterns.items():

        if re.search(pattern, message):

            matched_faqs.append(index)

    return matched_faqs
```

```python
print(recognize_intents("What are your working hours?"))

print(recognize_intents("How do I submit my internship project?"))

print(recognize_intents("When will I receive my certificate?"))
```

```python
message = "Tell me about internship project submission and certificate"

print(recognize_intents(message))
```

```python
def get_response(user_input):

    message = normalize_input(user_input)

    responses = []

    # Check greeting
    if is_greeting(message):

        responses.append(
            "Hello! Welcome to Student Support Assistant. How can I help you?"
        )

    # Recognize FAQ intents
    matched_faqs = recognize_intents(message)

    # Get answers from JSON dataset
    for index in matched_faqs:

        answer = faq_data["faqs"][index]["answer"]

        responses.append(answer)

    # Fallback response
    if not responses:

        return (
            "Sorry, I didn't understand your question. "
            "Please ask about business hours, internship submission, "
            "or certificates."
        )

    return "\n".join(responses)
```

```python
conversation_state = "greeting"

def chatbot(user_input):

    global conversation_state

    message = normalize_input(user_input)

    # Exit condition
    if message in ["bye", "exit", "quit"]:

        conversation_state = "exit"

        return "Thank you for using Student Support Assistant. Goodbye!"

    # Greeting state
    if conversation_state == "greeting":

        conversation_state = "conversation"

    # Generate chatbot response
    response = get_response(user_input)

    return response
```

```python
print("Current state:", conversation_state)

print(chatbot("Hello"))

print("Current state:", conversation_state)
```

```python
print(chatbot("bye"))

print("Current state:", conversation_state)
```

```python
conversation_state = "greeting"

print(chatbot("quit"))
```

```python
# Reset conversation state
conversation_state = "greeting"

print("=" * 50)
print("       STUDENT SUPPORT ASSISTANT")
print("=" * 50)

print("Bot: Hello! Welcome to Student Support Assistant.")
print("Bot: How can I help you today?")
print("Bot: Type 'bye', 'exit', or 'quit' to stop.")

while True:

    user_input = input("\nYou: ")

    response = chatbot(user_input)

    print("Bot:", response)

    if conversation_state == "exit":
        break
```
